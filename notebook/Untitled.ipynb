```python
import os
import pandas as pd
import numpy as np
import cv2
from tqdm import tqdm

# ===== FIXED PATHS =====
input_folder = "../dataset_csv"
output_folder = "../dataset_images"

os.makedirs(output_folder, exist_ok=True)

files = [f for f in os.listdir(input_folder) if f.endswith(".csv")]
files = sorted(files)

print(f"Found {len(files)} CSV files")

for file in tqdm(files):
    try:
        file_path = os.path.join(input_folder, file)

        data = pd.read_csv(
            file_path,
            header=None,
            sep=";",
            decimal=",",
            engine="python"
        ).values.astype(np.float32)

        img = (data - data.min()) / (data.max() - data.min() + 1e-8)
        img = (img * 255).astype(np.uint8)

        img_color = cv2.applyColorMap(img, cv2.COLORMAP_INFERNO)

        out_path = os.path.join(output_folder, file.replace(".csv", ".png"))
        cv2.imwrite(out_path, img_color)

    except Exception as e:
        print(f"❌ {file}: {e}")

print("✅ Done!")
```
