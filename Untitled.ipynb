```python
# run from thermal-cnn/
import pandas as pd, os

df = pd.read_parquet("dataset_V2.parquet")
df["png"] = df["IR_Image1Name"].str.replace(".csv", ".png", regex=False)
df["exists"] = df["png"].apply(lambda f: os.path.exists(f"dataset_images/{f}"))

print(df["exists"].value_counts())
print("\nMissing sample filenames:")
print(df[~df["exists"]]["IR_Image1Name"].head(10).tolist())
```

```python
import os, pandas as pd

df = pd.read_parquet("dataset_V2.parquet")
parquet_names = set(df["IR_Image1Name"].dropna().str.replace(".csv", ".png"))
disk_files    = set(os.listdir("dataset_images/"))

matched   = parquet_names & disk_files
unmatched = disk_files - parquet_names

print(f"Matched  : {len(matched)}")
print(f"On disk but NOT in parquet: {len(unmatched)}")
print("\nSample unmatched files on disk:")
print(list(unmatched)[:5])
```

```python
import os, pandas as pd

df = pd.read_parquet("dataset_V2.parquet")

# How many parquet rows are March 2022?
march = df[df["IR_Image1Name"].str.contains("20220317", na=False)]
print(f"Parquet rows for March 2022 : {len(march)}")
print(f"Of those, how many match disk: {march['IR_Image1Name'].str.replace('.csv','.png').apply(lambda f: os.path.exists(f'dataset_images/{f}')).sum()}")

# What experiment numbers are covered?
print(f"\nExperiment numbers in parquet (March):")
print(march["MET_ExperimentNumber"].value_counts())
```
